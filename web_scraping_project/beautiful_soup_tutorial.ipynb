## 5. HTML parsing with `Beautiful Soup`: A Mini Project

- scrape data from a public website
- organize the data as a dataframe
- export the data as an excel sheet

```python
url = 'https://www.bbc.com/sport/cricket/mens-england-division-one-multi-day/table'
```

```python
team_names = []
played = []
won = []
lost = []
drawn = []
points = []
```

```python
try:
    response = requests.get(url)
    response.raise_for_status()
except Exception as e:
    print(e)
else:
    soup = BeautifulSoup(response.content, 'html.parser')
    teams = soup.find('table').find_all('tr', class_='ssrcss-qqhdqi-TableRowBody e1icz100')

    for team in teams:
        team_name = team.find('div', class_='ssrcss-1m4x485-DesktopWidthOnly e1sk6csp2').get_text(strip=True)

        stats = team.find_all('div', class_='ssrcss-1vo7v3r-CellWrapper ef9ipf0')
        matches_played = int(stats[0].get_text(strip=True))
        matches_won = int(stats[2].get_text(strip=True))
        matches_lost = int(stats[4].get_text(strip=True))
        matches_drawn = int(stats[6].get_text(strip=True))
        total_points = int(team.find('div', class_='ssrcss-18ap757-CellWrapper ef9ipf0').get_text(strip=True))
        
        team_names.append(team_name)
        played.append(matches_played)
        won.append(matches_won)
        lost.append(matches_lost)
        drawn.append(matches_drawn)
        points.append(total_points)

    data = {
        'Team': team_names,
        'Played': played,
        'Won': won,
        'Lost': lost,
        'Drawn': drawn,
        'Points': points
    }
    df_teams = pd.DataFrame(data)
```
