<b>HEADERS</b>

```python
uri = "https://httpbin.org/headers"
```

```python
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)",
    "Accept": "application/json",
    "Authorization": "Bearer YOUR_ACCESS_TOKEN",
    "Content-Type": "application/json",
    "X-Custom-Header": "CustomValue"
}
```

```python
response = requests.get(uri, headers=headers)
```

```python
response.status_code
```

```python
response.headers
```

```python
type(response.headers)
```

```python
dict(response.headers)
```

```python
response.json()
```

```python
for header, value in response.json()["headers"].items():
    print(f"{header:<16}: {value}")
```

<b>Working with public API</b>

```python
uri = "https://jsonplaceholder.typicode.com/"
endpoint = "posts"
url = uri + endpoint
```

```python
try:
    response = requests.get(url)
    response.raise_for_status()
except Exception as e:
    print(e)
else:
    status_code = response.status_code
    print(f"Status Code: {status_code}")
    if status_code == 200:
        print("Successful GET request!")
        posts = response.json()
        for i in range(3):
            print(f"Post {i+1}: ")
            print(posts[i])
    else:
        print("Unsuccessful GET request!")
```

```python
new_post = {
    "title": "Sample Post",
    "body": "This is a sample post.",
    "userId": 101
}
```

```python
try:
    response = requests.post(url, data=new_post)
    response.raise_for_status()
except Exception as e:
    print(e)
else:
    status_code = response.status_code
    print(f"Status Code: {status_code}")
    if status_code == 201:
        print("Successful POST request!")
        post = response.json()
        print("\nPost:")
        print(post)
    else:
        print("Unsuccessful POST request!")
```
